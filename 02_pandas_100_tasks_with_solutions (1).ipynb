## Task 93: Use `pipe()` to apply a custom cleaning function.

**Category:** Intermediate Workflow & Export

```python
def clean_bank_data(data):
    return (data.replace(r'^\s*$', pd.NA, regex=True)
                .drop_duplicates(subset='customer_id'))
cleaned = pd.read_csv('../data/malaysia_banking_customers_500.csv').pipe(clean_bank_data)
cleaned.head()
```

## Task 94: Use `apply()` to create a risk comment for each customer.

**Category:** Intermediate Workflow & Export

```python
def risk_comment(row):
    if row['default_flag'] == 1 or row['credit_score'] < 580:
        return 'Review required'
    return 'Normal monitoring'
df['risk_comment'] = df.apply(risk_comment, axis=1)
df[['credit_score','default_flag','risk_comment']].head()
```
